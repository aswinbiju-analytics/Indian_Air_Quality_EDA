### Data Quality Assessment

The dataset was examined for missing values, duplicate records, data types, numerical distributions, and date validity.

Missing pollutant measurements were identified in the dataset. Instead of replacing these values with zero, a separate DataFrame was created for analyses requiring complete pollutant measurements. The original dataset was preserved to avoid unnecessary data loss.

Further cleaning decisions will depend on the observed data quality and the requirements of each analysis.

```python
df['Year'] = df['last_update'].dt.year
df['Month'] = df['last_update'].dt.month
df['Month_Name'] = df['last_update'].dt.month_name()
df['Hour'] = df['last_update'].dt.hour
df['Day_of_Week'] = df['last_update'].dt.day_name()

print("Time-based features created successfully!")

df[['last_update', 'Year', 'Month', 'Month_Name',
    'Hour', 'Day_of_Week']].head()
```

```python
print("AQI Summary:")
print(df['AQI'].describe())

print("\nMissing AQI Values:", df['AQI'].isnull().sum())
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='AQI',
    bins=30,
    kde=True
)

plt.title('Distribution of Air Quality Index (AQI)')
plt.xlabel('Air Quality Index (AQI)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Interpretation: AQI Distribution

The histogram shows how AQI values are distributed across the dataset. It helps identify the most frequent AQI ranges, the spread of observations, and potential extreme values.

A concentration of observations at lower AQI values indicates more records in those ranges, while a long tail toward higher values indicates fewer observations with high AQI readings. The actual pattern should be interpreted from the displayed chart.

Note: The dataset contains multiple locations and dates, so this distribution represents the combined records rather than the air quality of one specific city.

```python
print(df['AQI_Bucket'].value_counts(dropna=False))

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='AQI_Bucket',
    order=df['AQI_Bucket'].value_counts().index
)

plt.title('Distribution of AQI Categories')
plt.xlabel('AQI Category')
plt.ylabel('Number of Records')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Interpretation: AQI Category Distribution

This chart compares the number of records in each AQI category. The tallest bar represents the most frequently recorded category, while the shortest bar represents the least frequent category.

This analysis helps summarize the recorded air-quality classifications across the dataset. It does not, by itself, establish long-term trends or represent the current air quality of India.

```python
state_aqi = (
    df.groupby('state')['AQI']
    .mean()
    .sort_values(ascending=False)
)

print("Average AQI by State:")
display(state_aqi)
```

```python
top_10_states = state_aqi.head(10).sort_values()

plt.figure(figsize=(10, 6))

top_10_states.plot(kind='barh')

plt.title('Top 10 States by Average Recorded AQI')
plt.xlabel('Average AQI')
plt.ylabel('State')

plt.tight_layout()
plt.show()
```

### Interpretation: Average AQI by State

This chart compares the average recorded AQI across the ten states with the highest mean values in the dataset. The longest bar represents the state with the highest average AQI among these states.

The results help identify locations for further investigation. However, the comparison may be affected by differences in observation counts, stations, and dates. A higher AQI generally indicates poorer air quality, but these results alone do not establish the long-term air quality of an entire state.

```python
city_aqi = (
    df.groupby('city')['AQI']
    .mean()
    .sort_values(ascending=False)
)

top_10_cities = city_aqi.head(10).sort_values()

plt.figure(figsize=(10, 6))

top_10_cities.plot(kind='barh')

plt.title('Top 10 Cities by Average Recorded AQI')
plt.xlabel('Average AQI')
plt.ylabel('City')

plt.tight_layout()
plt.show()
```

### Interpretation: Average AQI by City

This chart identifies the ten cities with the highest average recorded AQI in the dataset. It helps compare the recorded air-quality levels across cities and highlights locations that may need closer examination.

These are dataset-level comparisons, not live AQI readings. Differences in monitoring coverage and dates may influence the results.

```python
season_aqi = pd.crosstab(
    df['Season'],
    df['AQI_Bucket']
)

display(season_aqi)

plt.figure(figsize=(10, 6))

sns.heatmap(
    season_aqi,
    annot=True,
    fmt='d',
    cmap='YlOrRd'
)

plt.title('AQI Category Distribution by Season')
plt.xlabel('AQI Category')
plt.ylabel('Season')

plt.tight_layout()
plt.show()
```

### Interpretation: AQI Category by Season

This heatmap compares the number of records in each AQI category across seasons. Darker cells indicate larger counts.

It helps identify seasonal differences in the recorded data. Because this chart displays counts rather than percentages, seasons with more observations may show higher counts even if their category proportions are similar.

```python

print("Unique Pollutants:")
print(df['pollutant_id'].value_counts(dropna=False))
```

```python
pollutant_summary = (
    df.groupby('pollutant_id')[
        ['pollutant_min', 'pollutant_max', 'pollutant_avg']
    ]
    .mean()
    .sort_values('pollutant_avg', ascending=False)
)

display(pollutant_summary)
```

```python
pollutant_avg = (
    df.groupby('pollutant_id')['pollutant_avg']
    .mean()
    .sort_values()
)

plt.figure(figsize=(10, 6))

pollutant_avg.plot(kind='barh')

plt.title('Average Recorded Measurement by Pollutant')
plt.xlabel('Average Pollutant Measurement')
plt.ylabel('Pollutant')

plt.tight_layout()
plt.show()
```

### Interpretation: Average Pollutant Measurements

This chart compares the mean recorded measurement for each pollutant. The longest bar represents the highest numerical average among the pollutants included.

Pollutant measurements may use different units and concentration scales, so this chart should not be used by itself to rank pollutants by health risk. It provides an initial comparison of recorded measurements in the dataset.

```python
numeric_columns = df.select_dtypes(
    include='number'
).columns.tolist()

print("Numerical Columns:")
print(numeric_columns)

correlation = df[numeric_columns].corr()

display(correlation)

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Matrix of Numerical Variables')
plt.tight_layout()
plt.show()
```

### Interpretation: AQI and Weather Correlation

The correlation heatmap displays the direction and strength of linear relationships between AQI and the selected weather variables.

Values closer to +1 indicate a stronger positive linear relationship, values closer to −1 indicate a stronger negative linear relationship, and values near zero indicate a weak linear relationship.

Correlation does not establish causation. These results describe associations within the dataset and do not prove that a weather variable causes a change in AQI.

```python
season_avg_aqi = (
    df.groupby('Season')['AQI']
    .mean()
    .sort_values(ascending=False)
)

print("Average AQI by Season:")
display(season_avg_aqi)

plt.figure(figsize=(9, 5))

season_avg_aqi.plot(kind='bar')

plt.title('Average AQI by Season')
plt.xlabel('Season')
plt.ylabel('Average AQI')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Interpretation: Average AQI by Season

This chart compares the average recorded AQI across seasons. The tallest bar represents the season with the highest mean AQI in the dataset.

Seasonal differences may be useful for identifying periods that deserve further investigation. However, the results depend on the dataset's observation dates, locations, and coverage. They do not establish that the season itself caused changes in air quality.

## Conclusion

This project conducted an Exploratory Data Analysis (EDA) of Indian air quality data using Python. The dataset contains 10,521 records covering 268 cities across 30 states.
